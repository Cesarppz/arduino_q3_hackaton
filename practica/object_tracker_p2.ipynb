```python
import cv2
import numpy as np
from PIL import Image

def detect_blue_object(img, threshold=250):
    image = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # maps
    maps = image.copy()
    blue_map = maps[:,:,-1]
    red_map = maps[:, :, 0]
    green_map = maps[:, :, 1]

    blue_mask = (blue_map >= 245) & (red_map <= 200) & (green_map <= 200)
    mask_white_and_black = Image.fromarray(blue_mask)
    bbox = mask_white_and_black.getbbox()

    if (bbox is not None) and (blue_mask.sum() > 300):
        x1, y1, x2, y2 = bbox

        # Calcular ancho (w) y alto (h) para el tracker de OpenCV
        w = x2 - x1
        h = y2 - y1

        object_detected = cv2.rectangle(
            image,
            pt1=(x1, y1),
            pt2=(x2, y2),
            color=(0,255,0),
            thickness=5)

        cx = int((x1 + x2) / 2)
        cy = int((y1 + y2) / 2)

        object_detected = cv2.circle(object_detected, (cx, cy), 7, (0, 0, 255), -1)

        # Guardar en formato [x, y, w, h]
        points = np.array([x1, y1, w, h])

    else:
        points = np.array([])
        object_detected = image

    return cv2.cvtColor(object_detected, cv2.COLOR_RGB2BGR), points

cam = cv2.VideoCapture(0)

while True:
    check, frame = cam.read()
    if not check:
        break

    new_frame, points_to_track = detect_blue_object(frame)

    cv2.imshow('video', new_frame)

    if points_to_track.any():
        # MedianFlow Tracker
        tracker = cv2.legacy.TrackerMedianFlow_create()
        # Convertir a tupla o lista plana, ya está en formato [x, y, w, h]
        ret = tracker.init(frame, tuple(points_to_track.tolist()))
        print('Objeto detectado, iniciando tracking...')
        break

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

while True:
    ret, frame = cam.read()
    if not ret:
        break

    success, roi = tracker.update(frame)

    if success:
        (x, y, w, h) = tuple(map(int, roi))
        p1 = (x, y)
        p2 = (x + w, y + h)
        cv2.rectangle(frame, p1, p2, (255, 0, 0), 2)
    else:
        cv2.putText(frame, "Fallo en el rastreo", (100, 80), cv2.FONT_HERSHEY_SIMPLEX, 0.75, (0, 0, 255), 2)

    # ESTA línea faltaba y causaba que el video pareciera congelado
    cv2.imshow('video', frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cam.release()
cv2.destroyAllWindows()
```

```python
# --- INICIO DEL PROGRAMA ---
cam = cv2.VideoCapture(1)
tracking_started = False

print("Fase 1: Buscando objeto azul (Presiona 'q' para salir)...")

# BUCLE 1: Fase de Detección
while True:
    check, frame = cam.read()
    if not check:
        print("Error leyendo la cámara.")
        break

    new_frame, points_to_track = detect_blue_object(frame)

    cv2.imshow('video', new_frame)

    if points_to_track.any():
        print("Objeto detectado. Inicializando VitTrack (Vision Transformer)...")

        try:
            # 1. Configurar los parámetros para VitTrack
            params = cv2.TrackerVit_Params()

            # Usamos el modelo estándar
            params.net = "object_tracking_vittrack_2023sep.onnx"

            # Si el video va lento, comenta la línea anterior y descomenta la siguiente
            # params.net = "object_tracking_vittrack_2023sep_int8bq.onnx"

            # 2. Crear el rastreador Vision Transformer
            tracker = cv2.TrackerVit_create(params)

            # 3. Inicializar con la caja delimitadora (convertida a tupla de enteros)
            bbox_tupla = tuple(map(int, points_to_track.tolist()))
            tracker.init(frame, bbox_tupla)

            tracking_started = True
            break # Salimos de la fase de detección para empezar el rastreo

        except cv2.error as e:
            print(f"Error de OpenCV: {e}")
            print("Asegúrate de que el archivo .onnx está en esta misma carpeta y tienes OpenCV >= 4.8.0")
            break

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# BUCLE 2: Fase de Rastreo por Red Neuronal (Solo se ejecuta si se detectó el objeto)
if tracking_started:
    print("Fase 2: Rastreo activo (Presiona 'q' para salir)...")

    while True:
        ret, frame = cam.read()
        if not ret:
            break

        # Actualizar la posición de la caja
        success, roi = tracker.update(frame)

        if success:
            # Dibujar la nueva caja delimitadora
            (x, y, w, h) = tuple(map(int, roi))
            p1 = (x, y)
            p2 = (x + w, y + h)
            cv2.rectangle(frame, p1, p2, (255, 0, 0), 2)
            cv2.putText(frame, "VitTrack Activo", (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 0, 0), 2)
        else:
            # Si el objeto desaparece de la pantalla
            cv2.putText(frame, "Rastreo perdido", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

        # Mostrar el resultado final
        cv2.imshow('video', frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

# Limpieza final
cam.release()
cv2.destroyAllWindows()
```
